## Diccionarios


```python
alumno = {
    "nombre": "Pepito",
    "edad": 32,
    "ciudad": "Granada"
}

print(alumno["nombre"])

alumno["matriculado"] = True
print(alumno)
alumno["nombre"] = "Pepe"
print(alumno)

###########

#Si intentamos acceder a yuna propiedad o clave sin que exista, nos arrojará un error por terminal
#print(alumno["altura"])

print(alumno.get("altura", "No exista la altura"))
print(alumno.get("nombre"))
```

```python
# Metodos de diccionarios
alumno = {
    "nombre": "Pepito",
    "edad": 32,
    "ciudad": "Granada"
}

# Keys -> Esto devuelve una vista de todas las claves
print(alumno.keys())
# Values -> Esto devuelve una lista de todos los valores
print(alumno.values())
# Items -> Devuelve una vista de pares en tuplas de clave y valor
print(alumno.items())
#Update -> actualizar un diccionario con otro diccionario
alumno.update({"altura":1.90, "pelo": "rubio"})
print(alumno)
# Pop -> le indicamos la clave y elimina la clave y el valor
alumno.pop("altura")
print(alumno)
#Copy -> hace una copia superficial del alumno
copia_alumno = alumno.copy()
copia_alumno["nombre"] = "Paco"
copia_alumno["edad"] = 25
print(copia_alumno)
```
