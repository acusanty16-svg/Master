## Operadores

```python
# Operadores aritmeticos
a = 11
b = 5

# Suma 
print(a+b)

# Resta 
print(a-b)

# Multipicacion
print(a*b)

# División
print(a/b)

# División entera 
print(a//b)

#Modulo
print(a%b)

#Potencia
print(a**b)
```

```python
# Operador de concatenacion

saludo = "Hola"
nombre = "Ana"
edad=30

mensaje = saludo + " " + nombre
segundo_mensaje = "Tengo " + str(edad) + " años"

print(mensaje)
print(segundo_mensaje)
```
