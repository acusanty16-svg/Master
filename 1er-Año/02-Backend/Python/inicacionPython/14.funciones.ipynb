## Funciones

```python
def saludar():
    print("Hola")

saludar()
saludar()
saludar()
```

```python
                    #Parametro
def saludar_con_nombre(nombre):
    print(f'Hola {nombre}')

                    #Argumento
saludar_con_nombre("Gabriel")
saludar_con_nombre("Sara")
```

```python
def sumar_dos_numeros(numeroA,numeroB):
    total = numeroA+numeroB
    print(total)

sumar_dos_numeros(5,6)
```

```python
def multiplicar_por_dos(numero):
    total= numero*2
    return total

resultado =multiplicar_por_dos(11) + multiplicar_por_dos(2)
print(resultado)
```

```python
def sumar(numeroa,numerob):
    """sumar dos numeros y retornar el resultado total

    Args:
        numeroa (int, float): Primer numero de la suma
        numerob (int, float): Segundo numero de la suma

    Returns:
        int, float: La suma de ambos numeros
    """
    return numeroa+numerob

sumar(5,6)
```

```python
# args -> parametro que recibe una cantidad indefinida/variables de valores
def mostrar_nombres(*args):
    for nombre in args:
        print(f'El nombre es: {nombre}')

mostrar_nombres("Paco","Gabriel","Alex","Carlos")

def sumar_numeros(*args):
    total=0
    for numero in args:
        total+=numero
    print(total)

sumar_numeros(1,2,3,4,5,20,12,190,45)
```

```python
#Kwargs -> recibir una cantidad indefinidad de claves y valores, generando asi un mini-diccionario

def mostrar_info(**kwargs):
    for clave, valor in kwargs.items():
        print(f'{clave} -> {valor}')

mostrar_info(nombre="Ana",edad=20,ciudad="Madrid",profesion="Carpintera")
```

```python
# lambda -> Funciones cortas reutilizables con retorno implicito
def sumar(numa, numb):
    return numa + numb

resultado_suma = lambda numa, numb: numa + numb
print(resultado_suma(2,5))
```
