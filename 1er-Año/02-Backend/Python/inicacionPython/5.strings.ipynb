## Strings

```python
texto = "Batman"

#Podemos indexar un string utlizando indices positivos o negativos para recorrerlo al derecho o al reves
print(texto[0])
print(texto[-1])
```

```python
# INDEZACION COMPLEJA
# string[start:stop:step]
# strar -> indice de inicio
# stop -> indice final
# step -> saltos entre caracteres

personaje ="Gandalf"

# extraer desde el 0 hasta el 3 -> el stop es excluido
print(personaje[0:3])
# Si no queremos indicar el start por obviedad podemos hacer lo siguiente:
print(personaje[:3])
# Empezamos desde 3 y omitimos el final
print(personaje[3:])
# Quiero leer el string entero pero con un salto de dos en dos caracteres
print(personaje[::2])
# Invertir un texto
print(personaje[::-1])
```

```python
# Metodos string
texto = "Ash"
print(len(texto))

# Lower
print(texto.lower())

#Upper
print(texto.upper())

frase = "el caballero oscuro"
#Title -> ponerle a cada uno de los substrings la primera en mayus
print(frase.title())

print("Clase de programacion con Python: el tipo de dato String".title())

#Capitalize -> la primera en mayus
print(frase.capitalize())

# Strip -> Elimina espacios incialies y finales
frase_con_espacios = "      Hola, buenas tardes.     "
print(len(frase_con_espacios))
frase_con_espacios = frase_con_espacios.strip()
print(len(frase_con_espacios))

# Replace -> reemplazar subtareas
frase_replace = "Buenas noches Gotham!"
print(frase_replace.replace("Gotham", "Móstoles"))

# Find -> Encuentra el primer indice del string o letra coincidente
frase= "Hola, que tal"
print(frase.find("x"))

#Index -> Hace lo mismo que find, pero si no lo encuentra arroja un error
# print(frase.index("x"))

# Join 
superheroes = ["Batman","Superman","Wonderwoman"]
print(" | ".join(superheroes))
```

```python
# Traductor a balleno
texto_original = input("Introduce un bloque de texto")
mapa = texto_original.maketrans("aeiuoAEIOU", "oooooOOOOO")
texto_balleno = texto_original.translate(mapa)
print(texto_balleno)
```
